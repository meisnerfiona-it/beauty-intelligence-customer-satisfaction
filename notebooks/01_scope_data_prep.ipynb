```python
# Right away I will drop any 'child' columns as these are not needed
products = products.drop(columns=["child_count", "child_max_price", "child_min_price"])

# Dropping these columns because most of their data is null
products = products.drop(columns=["value_price_usd", "sale_price_usd", "variation_desc", "variation_type", "variation_value"])

# Dropping rows with no rating/reviews
products = products.dropna(subset=["rating", "reviews"])
```

#### Reviews df

```python
all_reviews.head(2)
```

```python
# Before anything i need to make sure only the matching product ideas on the review and products dfs are matching
all_reviews = all_reviews.merge(skincare_products[["product_id"]],on="product_id",how="inner")
```

```python
all_reviews.head(5)
```

```python
all_reviews.shape
```

```python
all_reviews.isnull().sum()
```

```python
all_reviews['helpfulness'].value_counts(dropna=False).head(10)
```

```python
# Drop unnecessary index column
all_reviews = all_reviews.drop(columns="Unnamed: 0")
all_reviews = all_reviews.drop(columns="skin_tone")
all_reviews = all_reviews.drop(columns="hair_color")
all_reviews = all_reviews.drop(columns="eye_color")



# Remove reviews without review text
all_reviews = all_reviews.dropna(subset=["review_text"])

# Fill values that make sense
all_reviews["helpfulness"] = all_reviews["helpfulness"].fillna(0)
all_reviews["review_title"] = all_reviews["review_title"].fillna("No Title")

# Fill with uknown
all_reviews["skin_type"] = all_reviews["skin_type"].fillna("unknown")

# Date column
all_reviews['submission_time'] = pd.to_datetime(all_reviews['submission_time'], errors='coerce')
```

```python
all_reviews.groupby('skin_type')['rating'].mean()
```

```python
all_reviews["is_recommended"].value_counts(dropna=False)
```

```python
all_reviews["is_recommended"] = all_reviews["is_recommended"].fillna("unknown")
```

```python
all_reviews.duplicated().sum()
```

```python
duplicates = all_reviews[all_reviews.duplicated(keep=False)]
duplicates.sort_values("product_id").head(1)
```

```python
all_reviews = all_reviews.drop_duplicates()
```

```python
all_reviews.duplicated().sum()
```

```python
all_reviews.info()
```

```python
all_reviews["helpfulness"] = all_reviews["helpfulness"].astype(int)

# Need to drop for merge
all_reviews = all_reviews.drop(columns=["product_name", "brand_name", "price_usd", "rating"])
```

```python
# Review df is done being cleaned for now, saving as csv
all_reviews.to_csv("../data/processed/reviews_clean.csv")
```
